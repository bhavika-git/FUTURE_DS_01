```python
# Convert the Ship Date column from text (string) to a proper datetime format
df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    format="%m/%d/%Y"
)
# Check the data types of Order Date and Ship Date
# Both should now be datetime types
df[["Order Date", "Ship Date"]].dtypes
```

```python
# Display the unique values in the main categorical columns
# This helps us check for spelling differences, unexpected labels, or inconsistent categories

for column in ["Category", "Sub-Category", "Region", "Segment", "Ship Mode"]:
    print(f"\n{column}:")
    print(df[column].unique())
```

```python
# Negative profit values are important because they may reveal
# products or sales situations that are hurting business performance

loss_making_rows = (df["Profit"] < 0).sum()

print("Number of loss-making order lines:", loss_making_rows)
```

```python
# Calculate the number of loss-making order lines in each product category
# This helps us identify which major category has the most frequent losses

loss_by_category = (
    df[df["Profit"] < 0]
    .groupby("Category")
    .size()
    .sort_values(ascending=False)
)

print(loss_by_category)
```

```python
# Calculate the total profit from loss-making transactions for each category
# Since these values are negative, they represent the total financial loss

loss_amount_by_category = (
    df[df["Profit"] < 0]
    .groupby("Category")["Profit"]
    .sum()
    .sort_values()
)

print(loss_amount_by_category)
```

```python
# Calculate total Sales and Profit for each product category
# This lets us compare revenue generation with actual profitability

category_performance = (
    df.groupby("Category")[["Sales", "Profit"]]
    .sum()
    .sort_values("Sales", ascending=False)
)

print(category_performance)
```

```python
# Calculate total Sales and Profit for each sub-category
# This helps us identify the strongest and weakest product groups

subcategory_performance = (
    df.groupby("Sub-Category")[["Sales", "Profit"]]
    .sum()
    .sort_values("Sales", ascending=False)
)

print(subcategory_performance)
```

```python
# Calculate total Sales and Profit for each region
# This helps us compare business performance across different geographic areas

region_performance = (
    df.groupby("Region")[["Sales", "Profit"]]
    .sum()
    .sort_values("Sales", ascending=False)
)

print(region_performance)
```

```python
# Calculate total Sales and Profit for each customer segment
# This helps us understand which type of customer contributes most to business performance

segment_performance = (
    df.groupby("Segment")[["Sales", "Profit"]]
    .sum()
    .sort_values("Sales", ascending=False)
)

print(segment_performance)
```

```python
# Extract the year from Order Date
# We need this to compare business performance across different years

df["Year"] = df["Order Date"].dt.year

# Calculate total Sales and Profit for each year
# This helps us identify whether the business is growing or declining over time

yearly_performance = (
    df.groupby("Year")[["Sales", "Profit"]]
    .sum()
    .sort_index()
)

print(yearly_performance)
```

```python
# Create a Month column from Order Date
# Using the month period lets us group all transactions belonging to the same month

df["Month"] = df["Order Date"].dt.to_period("M")

# Calculate total Sales and Profit for each month
# This helps us identify growth patterns, seasonal peaks, and weaker periods

monthly_performance = (
    df.groupby("Month")[["Sales", "Profit"]]
    .sum()
    .sort_index()
)

print(monthly_performance)
```

```python
# Calculate total Sales for each individual product
# This helps us identify the products generating the most revenue

product_sales = (
    df.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

# Display the top 10 products by total Sales
print(product_sales.head(10))
```

```python
# Calculate total Profit for each individual product
# This helps us identify the products that contribute the most profit

product_profit = (
    df.groupby("Product Name")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

# Display the top 10 most profitable products
print(product_profit.head(10))
```

```python
# Calculate the total Profit generated by each individual product
# Sorting in ascending order puts the products with the biggest losses first

product_loss = (
    df.groupby("Product Name")["Profit"]
    .sum()
    .sort_values()
)

# Display the 10 products with the lowest total profit
# Negative values indicate that the product generated an overall loss

print(product_loss.head(10))
```

```python
# Calculate the average Profit for each Discount level
# This helps us investigate whether higher discounts are associated with lower profitability

discount_profit = (
    df.groupby("Discount")["Profit"]
    .agg(["mean", "sum", "count"])
    .sort_index()
)

print(discount_profit)
```

```python
# Check whether Sales or Quantity contains zero or negative values
# These checks help identify records that may need further investigation

invalid_sales = (df["Sales"] <= 0).sum()
invalid_quantity = (df["Quantity"] <= 0).sum()

print("Sales values <= 0:", invalid_sales)
print("Quantity values <= 0:", invalid_quantity)
```

```python
# Compare Order Date and Ship Date for every order line
# A valid shipment date should be the same as or later than the order date

invalid_ship_dates = (
    df["Ship Date"] < df["Order Date"]
).sum()

print("Ship dates earlier than order dates:", invalid_ship_dates)
```

```python
# Calculate the key business metrics for the entire dataset
# These metrics will later become the main KPI cards in our Power BI dashboard

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_quantity = df["Quantity"].sum()
total_orders = df["Order ID"].nunique()
total_customers = df["Customer ID"].nunique()

print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
print("Total Quantity Sold:", total_quantity)
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
```

```python
# Calculate the overall profit margin
# Profit margin shows what percentage of Sales is retained as Profit

profit_margin = (total_profit / total_sales) * 100

print("Overall Profit Margin:", profit_margin, "%")
```

```python
# Calculate Sales and Profit for each region
# We will use these values to calculate the profit margin for each region

region_metrics = (
    df.groupby("Region")[["Sales", "Profit"]]
    .sum()
)

# Calculate the percentage of Sales that becomes Profit in each region
region_metrics["Profit Margin (%)"] = (
    region_metrics["Profit"] / region_metrics["Sales"]
) * 100

# Sort regions by profit margin from highest to lowest
region_metrics = region_metrics.sort_values(
    "Profit Margin (%)",
    ascending=False
)

print(region_metrics)
```

```python
# Calculate Sales and Profit for each region
# We will use these values to calculate the profit margin for each region

region_metrics = (
    df.groupby("Region")[["Sales", "Profit"]]
    .sum()
)

# Calculate the percentage of Sales that becomes Profit in each region
region_metrics["Profit Margin (%)"] = (
    region_metrics["Profit"] / region_metrics["Sales"]
) * 100

# Sort regions by profit margin from highest to lowest
region_metrics = region_metrics.sort_values(
    "Profit Margin (%)",
    ascending=False
)

print(region_metrics)
```

```python
# Calculate total Profit for each product category
# Sorting from highest to lowest helps us identify the most profitable categories

category_profit = (
    df.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

print(category_profit)
```

```python
# Calculate total Sales and total Profit for each product category
# This helps us compare how much each category sells and how much profit it generates

category_summary = (
    df.groupby("Category")[["Sales", "Profit"]]
    .sum()
    .sort_values("Sales", ascending=False)
)

# Display the category-wise Sales and Profit
print(category_summary)
```

```python
# Rank regions by their total Profit
# This helps us identify which geographic region contributes the most profit

region_profit = (
    df.groupby("Region")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

print(region_profit)
```

```python
# Rank all regions by their total Profit
# This helps us identify which geographic region contributes the most
# profit to the overall business

region_profit = (
    df.groupby("Region")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

print(region_profit)
```

```python
# Calculate total Sales and Profit for each sub-category
# This gives us a more detailed view of which product groups
# are performing strongly or poorly within each major category

subcategory_metrics = (
    df.groupby("Sub-Category")[["Sales", "Profit"]]
    .sum()
    .sort_values("Profit", ascending=False)
)

# Display all sub-categories ranked from highest to lowest profit
print(subcategory_metrics)
```

```python
# Calculate total Sales and Profit for each sub-category
# We need both values to determine how efficiently each sub-category
# converts its Sales into Profit

subcategory_margin = (
    df.groupby("Sub-Category")[["Sales", "Profit"]]
    .sum()
)

# Calculate Profit Margin as a percentage of Sales
# A higher percentage means the sub-category keeps more of its
# revenue as profit

subcategory_margin["Profit Margin (%)"] = (
    subcategory_margin["Profit"] / subcategory_margin["Sales"]
) * 100

# Rank sub-categories from highest to lowest profit margin
# This helps us identify the most efficient product groups

subcategory_margin = subcategory_margin.sort_values(
    "Profit Margin (%)",
    ascending=False
)

print(subcategory_margin)
```

```python
# Filter the sub-categories where total Profit is negative
# These are product groups that lose money overall
# and therefore deserve business attention

loss_making_subcategories = subcategory_metrics[
    subcategory_metrics["Profit"] < 0
].sort_values("Profit")

# Display the loss-making sub-categories
print(loss_making_subcategories)
```

```python
# Calculate total Sales and Profit for each region
# We need both values so we can measure how profitable each region is

regional_summary = (
    df.groupby("Region")[["Sales", "Profit"]]
    .sum()
)

# Calculate Profit Margin for each region
# This tells us what percentage of Sales is converted into Profit

regional_summary["Profit Margin (%)"] = (
    regional_summary["Profit"] / regional_summary["Sales"]
) * 100

# Sort regions by Profit Margin from highest to lowest
# This makes it easy to identify the most and least efficient regions

regional_summary = regional_summary.sort_values(
    "Profit Margin (%)",
    ascending=False
)

print(regional_summary)
```
